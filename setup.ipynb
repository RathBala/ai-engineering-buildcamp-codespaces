# Buildcamp setup check

Run all cells with the project's .venv Python environment. The API key comes from the OPENAI_API_KEY Codespaces secret. Never paste it into this notebook.

The first two checks verify the environment and the lesson's OpenAI client setup. The final check verifies authentication with a read-only models request. It does not generate text or use model tokens.

```python
from pathlib import Path
import sys

assert Path(sys.prefix).resolve() == (Path.cwd() / '.venv').resolve(), 'Select this project’s .venv Python environment.'
print('Project environment: OK')
```

```python
from openai import OpenAI

client = OpenAI()
print('OpenAI client setup: OK')
```

```python
from openai import APIConnectionError, APIStatusError

try:
    client.with_options(timeout=20.0, max_retries=0).models.list()
except APIStatusError as error:
    raise RuntimeError(f'OpenAI authentication check failed (HTTP {error.status_code}). Check the course key and its permissions.') from None
except APIConnectionError:
    raise RuntimeError('Could not reach OpenAI. Check the connection and retry.') from None
print('OpenAI authentication: OK')
```
